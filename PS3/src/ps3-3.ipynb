(a)

$$
\begin{align}
E &= \int p(y;\theta) \nabla_\theta \log p(y;\theta)dy
\\
&= \int p(y;\theta) \frac{1}{p(y;\theta)} \nabla_\theta p(y;\theta)dy
\\
&= \int \nabla_\theta p(y;\theta)dy
\\
&= \nabla_\theta \int p(y;\theta)dy
\\
&= \nabla_\theta 1
\\
&=0
\end{align}
$$

(b)

For a random vector $X$, the covariance matrix is

$$ Cov(X,X) = E[(X-E(X))(X-E(X))^T]$$ 

If $E(X) = 0$, then 

$$ Cov(X,X) = E[XX^T]$$ 

If we let $X = \nabla_\theta \log p(y;\theta)$, we have proved in (a) that $E(X)=0$. So 

$$\mathcal{I}(\theta) = Cov_{y\sim p(y;\theta)} [\nabla_\theta \log p(y;\theta)] = E_{y\sim p(y;\theta)}[\nabla_\theta \log p(y;\theta) \nabla_\theta \log p(y;\theta)^T]$$

(c)

$$\mathcal{I}(\theta)_{ij} = E_{y\sim p(y;\theta)}\left[\frac{1}{p(y;\theta)^2}\frac{\partial p(y;\theta)}{\partial \theta_i}\frac{\partial p(y;\theta)}{\partial \theta_j}\right]$$

$$H_{ij} = \frac{\partial^2}{\partial \theta_i \partial \theta_j} \log p(y;\theta) = \frac{\partial}{\partial \theta_j}\left(\frac{1}{p(y;\theta)}\frac{\partial p(y;\theta)}{\partial \theta_i} \right) = -\frac{1}{p(y;\theta)^2}\frac{\partial p(y;\theta)}{\partial \theta_i}\frac{\partial p(y;\theta)}{\partial \theta_j} + \frac{1}{p(y;\theta)}\frac{\partial^2 p(y;\theta)}{\partial \theta_i \partial \theta_j} $$

So 

$$
\begin{align}
E_{y\sim p(y;\theta)}[-H_{ij}] &= E_{y\sim p(y;\theta)}\left[\frac{1}{p(y;\theta)^2}\frac{\partial p(y;\theta)}{\partial \theta_i}\frac{\partial p(y;\theta)}{\partial \theta_j}\right] + E_{y\sim p(y;\theta)} \left[\frac{1}{p(y;\theta)}\frac{\partial^2 p(y;\theta)}{\partial \theta_i \partial \theta_j} \right]
\end{align}
$$

The second term on the right hand side is 0 because

$$\int p(y;\theta) \frac{1}{p(y;\theta)}\frac{\partial^2 p(y;\theta)}{\partial \theta_i \partial \theta_j} dy = \frac{\partial^2 }{\partial \theta_i \partial \theta_j} \int p(y;\theta) dy = \frac{\partial^2 }{\partial \theta_i \partial \theta_j} 1 = 0$$

Therefore 

$$E_{y\sim p(y;\theta)}[-H_{ij}] = \mathcal{I}(\theta)_{ij}$$

Or

$$E_{y\sim p(y;\theta)}[-H] = E_{y\sim p(y;\theta)}[-\nabla_\theta^2 \log p(y;\theta)] = \mathcal{I}(\theta)$$

(d)

$$D_{KL}(p_\theta||p_{\tilde \theta}) = \int p(y;\theta) \log \frac{p(y;\theta)}{p(y;\tilde \theta)}dy = \int p(y;\theta) \log p(y;\theta)dy - \int p(y;\theta) \log p(y;\tilde \theta)dy$$

From the Taylor explansion

$$f(\tilde \theta) \approx f(\theta) + (\tilde \theta - \theta)^T \nabla_{\theta'}f(\theta')|_{\theta'=\theta} + \frac{1}{2}(\tilde \theta - \theta)^T (\nabla_{\theta'}^2f(\theta')|_{\theta'=\theta}) (\tilde \theta - \theta)$$

Let $f(\theta) = \log p(y;\theta)$, we have

$$\log p(y;\tilde \theta) \approx \log p(y;\theta) + (\tilde \theta - \theta)^T \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} + \frac{1}{2}(\tilde \theta - \theta)^T (\nabla_{\theta'}^2\log p(y;\theta')|_{\theta'=\theta}) (\tilde \theta - \theta)$$

So

$$
\begin{align}
D_{KL}(p_\theta||p_{\tilde \theta}) 
&\approx -\int p(y;\theta) (\tilde \theta - \theta)^T \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \; dy - \int p(y;\theta) \frac{1}{2}(\tilde \theta - \theta)^T (\nabla_{\theta'}^2\log p(y;\theta')|_{\theta'=\theta}) (\tilde \theta - \theta) dy
\\
&= -(\tilde \theta - \theta)^T \int p(y;\theta)  \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \; dy + \frac{1}{2}(\tilde \theta - \theta)^T \left( \int p(y;\theta)  (-\nabla_{\theta'}^2\log p(y;\theta')|_{\theta'=\theta}) dy \right) (\tilde \theta - \theta)
\\
&= -(\tilde \theta - \theta)^T E_{y\sim p(y;\theta)}[\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta}] + \frac{1}{2}(\tilde \theta - \theta)^T E_{y\sim p(y;\theta)}[-\nabla_{\theta'}^2\log p(y;\theta')|_{\theta'=\theta}](\tilde \theta - \theta)
\end{align}
$$

Using the result from (a) and (c), we have

$$D_{KL}(p_\theta||p_{\tilde \theta}) \approx \frac{1}{2}(\tilde \theta - \theta)^T \mathcal{I}(\theta) (\tilde \theta - \theta)$$

Setting $\tilde \theta = \theta + d$, we have

$$D_{KL}(p_\theta||p_{\theta + d}) \approx \frac{1}{2}d^T \mathcal{I}(\theta) d$$

(e)

The optimization problem is

$$d^* = \arg \max_d \ell(\theta+d) \approx  \arg \max_d d^T\nabla_{\theta'} \ell(\theta')|_{\theta'=\theta}$$

subject to

$$D_{KL}(p_\theta||p_{\theta + d}) \approx \frac{1}{2}d^T \mathcal{I}(\theta) d = c $$

Therefore the Lagrangian is

$$\mathcal{L}(d,\lambda) = d^T\nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} - \lambda (\frac{1}{2}d^T \mathcal{I}(\theta) d - c)$$

$$Setting $\nabla_d \mathcal{L} = 0$ and $\nabla_\lambda \mathcal{L} = 0$, we have

$$\nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} - \lambda \mathcal{I}(\theta) d = 0$$

$$\frac{1}{2}d^T \mathcal{I}(\theta) d - c = 0$$

From the first equation above we can express $d$ as

$$d=\frac{1}{\lambda} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} $$

So

$$\tilde d = \frac{1}{\lambda} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} $$

We can solve for $\lambda$ by substituting the expression of $d$ into the other equation

$$\frac{1}{\lambda^2} \nabla_{\theta'} \ell(\theta')^T|_{\theta'=\theta}  \mathcal{I}(\theta)^{-1} \mathcal{I}(\theta) \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} =2c$$

So

$$\lambda = \sqrt{\frac{\nabla_{\theta'} \ell(\theta')^T|_{\theta'=\theta} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} }{2c}}$$

Plug it back into the expression of $d$, we have

$$d=\sqrt{\frac{2c}{\nabla_{\theta'} \ell(\theta')^T|_{\theta'=\theta} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} }} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} $$

So

$$d^*=\sqrt{\frac{2c}{\nabla_{\theta'} \ell(\theta')^T|_{\theta'=\theta} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} }} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} $$

But remember $ \ell(\theta) = \log p(y;\theta)$. So

$$\tilde d = \frac{1}{\lambda} \mathcal{I}(\theta)^{-1}  \frac{\nabla_{\theta'} p(y;\theta')|_{\theta'=\theta}}{p(y;\theta)} $$

$$d^*=\sqrt{\frac{2c p(y;\theta)^2}{\nabla_{\theta'} p(y;\theta')^T|_{\theta'=\theta} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} p(y;\theta')|_{\theta'=\theta} }} \mathcal{I}(\theta)^{-1} \frac{\nabla_{\theta'} p(y;\theta')|_{\theta'=\theta}}{p(y;\theta)} = \sqrt{\frac{2c}{\nabla_{\theta'} p(y;\theta')^T|_{\theta'=\theta} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} p(y;\theta')|_{\theta'=\theta} }} \mathcal{I}(\theta)^{-1} \nabla_{\theta'} p(y;\theta')|_{\theta'=\theta}$$

(f)
