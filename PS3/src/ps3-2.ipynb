(a)

$$
\begin{align}
D_{KL}P(P||Q) &= \sum_{x \in \chi} P(x) \log\frac{P(x)}{Q(x)}
\\
&= \sum_{x \in \chi} P(x) [-\log\frac{Q(x)}{P(x)}]
\\
&= E [-\log\frac{Q(x)}{P(x)}]
\\
&\geq -\log \left(E [\frac{Q(x)}{P(x)}] \right)
\\
&= -\log \left(\sum_{x \in \chi} P(x) \frac{Q(x)}{P(x)} \right)
\\
&= -\log \left(\sum_{x \in \chi} Q(x) \right)
\\
&= -\log 1
\\
&=0
\end{align}$$

To prove $D_{KL}P(P||Q) = 0$ if and only if $P=Q$, we just need to prove they are equivalent.

It is easy to see that 

$$P=Q \rightarrow D_{KL}P(P||Q) = 0$$

We then just need to prove the other direction. From $D_{KL}P(P||Q) = 0$ we have (from the derivation above)

$$ E[-\log\frac{Q(x)}{P(x)}] = -\log \left(E [\frac{Q(x)}{P(x)}] \right)$$

since $f(x) = -\log(x)$ is strictly convex, from Jensen's inequality we have

$$\frac{Q(x)}{P(x)} = E(\frac{Q(x)}{P(x)})$$

with probability 1. 

Notice $E(\frac{Q(x)}{P(x)}) = \sum_{x \in \chi} P(x) \frac{Q(x)}{P(x)} = 1$, so $P(x) = Q(x)$ with probability 1, aka they are equal.

(b)

$$
\begin{align}
D_{KL}(P(Y|X)||Q(Y|X)) &= \sum_x P(x)\left( \sum_y P(y|x) \log\frac{P(y|x)}{Q(y|x)} \right)
\\
&= \sum_x \sum_y P(x)P(y|x) \log \left( \frac{P(x,y)}{Q(x,y)} \frac{Q(x)}{P(x)} \right)
\\
&= \sum_x \sum_y P(x,y) \left (\log \frac{P(x,y)}{Q(x,y)} + \log \frac{Q(x)}{P(x)} \right)
\\
&= \sum_x \sum_y P(x,y) \log \frac{P(x,y)}{Q(x,y)} - \sum_x \sum_y P(x,y) \log \frac{P(x)}{Q(x)}
\\
&= \sum_x \sum_y P(x,y) \log \frac{P(x,y)}{Q(x,y)} - \sum_x P(x) \log \frac{P(x)}{Q(x)}
\\
&= D_{KL}(P(X,Y)||Q(X,Y)) - D_{KL}(P(X)||Q(X))
\end{align}
$$

So

$$ D_{KL}(P(X,Y)||Q(X,Y)) = D_{KL}(P(X)||Q(X)) + D_{KL}(P(Y|X)||Q(Y|X))$$

(c)

Since $x$ is continuous, we have

$$
\begin{align}
D_{KL}(\hat P || P_\theta) &= \int_{x \in \chi} \hat P(x) \log \frac{\hat P(x)}{P_\theta(x)}
\\
&= \frac{1}{m} \sum_{i=1}^m \log \frac{1}{P_\theta(x^{(i)})}
\\
&= -\frac{1}{m} \sum_{i=1}^m \log P_\theta(x^{(i)})
\end{align}
$$

Therefore,

$$\arg \min_\theta D_{KL}(\hat P || P_\theta) = \arg \min_\theta  -\frac{1}{m} \sum_{i=1}^m \log P_\theta(x^{(i)}) = \arg \max_\theta \sum_{i=1}^m \log P_\theta(x^{(i)})$$
