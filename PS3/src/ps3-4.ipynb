(a)

We start from the definition

$$
\ell_{semi-sup}(\theta) = \ell_{unsup}(\theta) + \alpha \ell_{sup}(\theta) = \sum_{i=1}^m \log \sum_{z^{(i)}} p(x^{(i)}, z^{(i)};\theta) + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)
$$

For any distribution $Q(z)$ over the possible values of $z$ $\left( \sum_z Q(z) = 1, Q(z) \geq 0 \right)$, we have
$$
\begin{align}
\ell_{semi-sup}(\theta) &= \sum_{i=1}^m \log \sum_{z^{(i)}} p(x^{(i)}, z^{(i)};\theta) + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)
\\
&= \sum_{i=1}^m \log \sum_{z^{(i)}} Q_i(z^{(i)}) \frac{p(x^{(i)}, z^{(i)};\theta)}{Q_i(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)
\\
&= \sum_{i=1}^m \log E\left[\frac{p(x^{(i)}, z;\theta)}{Q_i(z)}\right] + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)
\\
&\geq \sum_{i=1}^m  E\left[\log \frac{p(x^{(i)}, z;\theta)}{Q_i(z)}\right] + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)
\\
&= \sum_{i=1}^m \sum_{z^{(i)}} Q_i(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)};\theta)}{Q_i(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)
\end{align}
$$

This means 

$$\ell_{semi-sup}(\theta) \geq \sum_{i=1}^m \sum_{z^{(i)}} Q_i(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)};\theta)}{Q_i(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)$$

for any $Q_i(z)$. Therefore, we have

$$\ell_{semi-sup}(\theta^{(t+1)}) \geq \sum_{i=1}^m \sum_{z^{(i)}} Q_i^{(t)}(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)};\theta^{(t+1)})}{Q_i^{(t)}(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta^{(t+1)})$$

From the M-step we have

$$\theta^{(t+1)} = \arg \max_\theta \sum_{i=1}^m \sum_{z^{(i)}} Q_i^{(t)}(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)};\theta)}{Q_i^{(t)}(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)$$

So 

$$\sum_{i=1}^m \sum_{z^{(i)}} Q_i^{(t)}(z^{(i)}) \log \frac{p(x^{(i)}, z;\theta^{(t+1)})}{Q_i^{(t)}(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta^{(t+1)}) \geq \sum_{i=1}^m \sum_{z^{(i)}} Q_i^{(t)}(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)};\theta^{(t)})}{Q_i^{(t)}(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta^{(t)})$$

Therefore

$$\ell_{semi-sup}(\theta^{(t+1)}) \geq \sum_{i=1}^m \sum_{z^{(i)}} Q_i^{(t)}(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)};\theta^{(t)})}{Q_i^{(t)}(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta^{(t)}) = \ell_{semi-sup}(\theta^{(t)}) $$

(b)

The E-step is identical to the unsupervised case.

$$
\begin{align}
w_j^{(i)} &= Q_i(z^{(i)}=j) = p(z^{(i)}=j | x^{(i)};\phi,\mu,\Sigma) 
\\
&= \frac{p(x^{(i)} | z^{(i)} = j;\mu,\Sigma)p(z^{(i)} =j;\phi)}{\sum_{l=1}^k p(x^{(i)} | z^{(i)} = l;\mu,\Sigma)p(z^{(i)} =l;\phi)}
\\
&= \frac{\frac{1}{(2\pi)^{n/2} |\Sigma_j|^{1/2}} \exp(-\frac{1}{2}(x^{(i)}-\mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j))\phi_j}{\sum_{l=1}^k \frac{1}{(2\pi)^{n/2} |\Sigma_l|^{1/2}} \exp(-\frac{1}{2}(x^{(i)}-\mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l))\phi_l}
\end{align} 
$$

(c)

In the M-step, the optimization objective is the lower bound of $\ell_{semi-sup}$:

$$
\begin{align}
&\sum_{i=1}^m \sum_{z^{(i)}} Q_i(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)};\theta)}{Q_i(z^{(i)})} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}, \tilde z^{(i)};\theta)
\\
=& \sum_{i=1}^m \sum_{j=1}^k Q_i(z^{(i)}=j) \log \frac{p(x^{(i)}|z^{(i)}=j;\theta)p(z^{(i)}=j;\theta)}{Q_i(z^{(i)}=j)} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}| \tilde z^{(i)};\theta)p(\tilde z^{(i)};\theta)
\\
=& \sum_{i=1}^m \sum_{j=1}^k w_j^{(i)} \log \frac{p(x^{(i)}|z^{(i)}=j;\theta)p(z^{(i)}=j;\theta)}{w_j^{(i)}} + \alpha \sum_{i=1}^{\tilde m} \log p(\tilde x^{(i)}| \tilde z^{(i)};\theta)+ \log p(\tilde z^{(i)};\theta)
\end{align}
$$

For GMM, this optimization objective reduces to

$$\sum_{i=1}^m \sum_{j=1}^k w_j^{(i)} \log \frac{\frac{1}{(2\pi)^{n/2} |\Sigma_j|^{1/2}} \exp(-\frac{1}{2}(x^{(i)}-\mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j))\phi_j}{w_j^{(i)}} + \alpha \sum_{i=1}^{\tilde m}\left[ \log \frac{1}{(2\pi)^{n/2} |\Sigma_{\tilde z^{(i)}}|^{1/2}} \exp(-\frac{1}{2}(\tilde x^{(i)}-\mu_{\tilde z^{(i)}})^T \Sigma_{\tilde z^{(i)}}^{-1} (\tilde x^{(i)} - \mu_{\tilde z^{(i)}})) + \log \phi_{\tilde z^{(i)}} \right]$$

Take a gradient of the optimization object with respect to $\mu_j$:

$$
\begin{align}
\nabla_{\mu_j} (...) &= \nabla_{\mu_j} \sum_{i=1}^m w_j^{(i)} (-\frac{1}{2})(x^{(i)}-\mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) + \alpha \nabla_{\mu_j} \sum_{i=1}^{\tilde m} -\frac{1}{2}(\tilde x^{(i)}-\mu_{\tilde z^{(i)}})^T \Sigma_{\tilde z^{(i)}}^{-1} (\tilde x^{(i)} - \mu_{\tilde z^{(i)}})
\\
&= \sum_{i=1}^m w_j^{(i)} \Sigma_j^{-1} (x^{(i)} - \mu_j) + \alpha \sum_{i=1}^{\tilde m} 1\{\tilde z^{(i)} = j\} \Sigma_j^{-1} (x^{(i)} - \mu_j)
\end{align}
$$

Setting the gradient to 0, we have

$$\mu_j = \frac{\sum_{i=1}^m w_j^{(i)} x^{(i)} + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\} \tilde x^{(i)}}{\sum_{i=1}^m w_j^{(i)} + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\}}$$

Then we optimize with respect to $\phi_j$. Notice that we have a constraint

$$\sum_{j=1}^k \phi_j = 1$$

So we should use Lagrange multiplier and add $\lambda (\sum_{j=1}^k \phi_j - 1)$ to the optimization objective. Then

$$
\begin{align}
\frac{\partial}{\partial \phi_j}\left(...+ \lambda (\sum_{j=1}^k \phi_j - 1)\right) = \frac{\partial}{\partial \phi_j} \left(\sum_{i=1}^m w_j^{(i)} \log \phi_j + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\} \log \phi_j + \lambda \phi_j \right) 
\end{align}
$$

Setting the derivative to 0, we have

$$\sum_{i=1}^m w_j^{(i)} \frac{1}{\phi_j} + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\} \frac{1}{\phi_j} + \lambda = 0$$

So 

$$\phi_j = \frac{\sum_{i=1}^m w_j^{(i)} + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\}}{-\lambda}$$

Plug this into the constraint, we have

$$\sum_{j=1}^k \frac{\sum_{i=1}^m w_j^{(i)} + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\}}{-\lambda} = 1$$

So 

$$
\begin{align}
-\lambda &= \sum_{j=1}^k \left( \sum_{i=1}^m w_j^{(i)} + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\} \right)
\\
&= \sum_{i=1}^m \sum_{j=1}^k w_j^{(i)} + \alpha \sum_{i=1}^{\tilde m} \sum_{j=1}^k 1\{ \tilde z^{(i)} = j\}
\\
&= \sum_{i=1}^m 1 + \alpha \sum_{i=1}^{\tilde m} 1
\\
&= m + \alpha \tilde m
\end{align}
$$

Therefore we have

$$\phi_j = \frac{\sum_{i=1}^m w_j^{(i)} + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\}}{m + \alpha \tilde m}$$

Finally, let use calculate the gradient of the optimization objective with respect to $\Sigma_j$.

$$
\begin{align}
\nabla_{\Sigma_j} (...) = \nabla_{\Sigma_j} \left[\sum_{i=1}^m w_j^{(i)} (-\log|\Sigma_j|^{1/2} -\frac{1}{2}(x^{(i)}-\mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j)) + \alpha \sum_{i=1}^{\tilde m} (-\log|\Sigma_{z^{(i)}}|^{1/2} -\frac{1}{2}(\tilde x^{(i)}-\mu_{z^{(i)}})^T \Sigma_{z^{(i)}}^{-1} (\tilde x^{(i)} - \mu_{z^{(i)}})) \right]
\end{align}
$$

Using the properties 

$$\nabla_\Sigma \log|\Sigma| = \Sigma^{-1}$$

for a symmetric matrix $\Sigma$, and

$$\nabla_\Sigma(a^T \Sigma^{-1} a) = -\Sigma^{-1} aa^T \Sigma^{-1}$$

We have

$$
\nabla_{\Sigma_j} (...) = \sum_{i=1}^m w_j^{(i)} (-\frac{1}{2} \Sigma_j^{-1} +\frac{1}{2}\Sigma_j^{-1} (x^{(i)}-\mu_j) (x^{(i)} - \mu_j)^T \Sigma_j^{-1}) + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\}(-\frac{1}{2} \Sigma_j^{-1} +\frac{1}{2}\Sigma_j^{-1} (\tilde x^{(i)}-\mu_j) (\tilde x^{(i)} - \mu_j)^T \Sigma_j^{-1})
$$

Setting the gradient to 0, we have

$$
\sum_{i=1}^m w_j^{(i)} \left(- \Sigma_j + (x^{(i)}-\mu_j) (x^{(i)} - \mu_j)^T \right) + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\} \left(- \Sigma_j +(\tilde x^{(i)}-\mu_j) (\tilde x^{(i)} - \mu_j)^T \right) = 0
$$

We can solve for $\Sigma_j$

$$\Sigma_j = \frac{\sum_{i=1}^m w_j^{(i)} (x^{(i)}-\mu_j) (x^{(i)} - \mu_j)^T  + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\} (\tilde x^{(i)}-\mu_j) (\tilde x^{(i)} - \mu_j)^T}{\sum_{i=1}^m w_j^{(i)}  + \alpha \sum_{i=1}^{\tilde m} 1\{ \tilde z^{(i)} = j\}}$$
